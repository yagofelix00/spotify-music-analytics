## 9. Track duration analysis

```python
tracks["duration_minutes"].describe()
```

```python
shortest_tracks = (
    tracks[
        [
            "name",
            "artist_name",
            "album_name",
            "duration_minutes",
        ]
    ]
    .sort_values(
        by="duration_minutes",
        ascending=True,
    )
    .head(10)
)

shortest_tracks
```

```python
longest_tracks = (
    tracks[
        [
            "name",
            "artist_name",
            "album_name",
            "duration_minutes",
        ]
    ]
    .sort_values(
        by="duration_minutes",
        ascending=False,
    )
    .head(10)
)

longest_tracks
```

```python
duration_bins = pd.cut(
    tracks["duration_minutes"],
    bins=[
        0,
        2,
        3,
        4,
        5,
        7,
        float("inf"),
    ],
    labels=[
        "< 2 min",
        "2–3 min",
        "3–4 min",
        "4–5 min",
        "5–7 min",
        "> 7 min",
    ],
    right=False,
)

duration_distribution = (
    duration_bins
    .value_counts()
    .sort_index()
)

duration_distribution
```

```python
duration_distribution_percentage = (
    duration_bins
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
)

duration_distribution_percentage
```

```python
duration_by_artist = (
    tracks.groupby("artist_name")["duration_minutes"]
    .agg(
        [
            "count",
            "mean",
            "median",
            "std",
            "min",
            "max",
        ]
    )
    .sort_values(
        by="mean",
        ascending=False,
    )
)

duration_by_artist
```

```python
q1 = tracks["duration_minutes"].quantile(0.25)
q3 = tracks["duration_minutes"].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

lower_bound, upper_bound
```

```python
duration_outliers = tracks[
    (tracks["duration_minutes"] < lower_bound)
    | (tracks["duration_minutes"] > upper_bound)
][
    [
        "name",
        "artist_name",
        "album_name",
        "duration_minutes",
    ]
].sort_values(
    by="duration_minutes",
    ascending=False,
)

duration_outliers
```

```python
len(duration_outliers)
```

```python
duration_analysis_summary = pd.Series(
    {
        "average_duration_min": tracks["duration_minutes"].mean(),
        "median_duration_min": tracks["duration_minutes"].median(),
        "standard_deviation_min": tracks["duration_minutes"].std(),
        "shortest_track_min": tracks["duration_minutes"].min(),
        "longest_track_min": tracks["duration_minutes"].max(),
        "q1": q1,
        "q3": q3,
        "iqr": iqr,
        "lower_outlier_bound": lower_bound,
        "upper_outlier_bound": upper_bound,
        "outlier_count": len(duration_outliers),
    }
)

duration_analysis_summary
```

## 10. Explicit content analysis

```python
explicit_summary = (
    tracks["explicit"]
    .value_counts()
    .rename_axis("explicit")
    .reset_index(name="track_count")
)

explicit_summary
```

```python
explicit_summary["percentage"] = (
    explicit_summary["track_count"]
    / explicit_summary["track_count"].sum()
    * 100
)

explicit_summary
```

```python
explicit_by_artist = (
    tracks.groupby("artist_name")
    .agg(
        total_tracks=("spotify_id", "count"),
        explicit_tracks=("explicit", "sum"),
    )
    .reset_index()
)
```

```python
explicit_by_artist["non_explicit_tracks"] = (
    explicit_by_artist["total_tracks"]
    - explicit_by_artist["explicit_tracks"]
)
```

```python
explicit_by_artist["explicit_percentage"] = (
    explicit_by_artist["explicit_tracks"]
    / explicit_by_artist["total_tracks"]
    * 100
)
```

```python
explicit_by_artist = explicit_by_artist.sort_values(
    by="explicit_percentage",
    ascending=False,
)

explicit_by_artist
```

```python
duration_by_explicit = (
    tracks.groupby("explicit")["duration_minutes"]
    .agg(
        [
            "count",
            "mean",
            "median",
            "std",
            "min",
            "max",
        ]
    )
)

duration_by_explicit
```

```python
explicit_artist_duration = (
    tracks.groupby(
        [
            "artist_name",
            "explicit",
        ]
    )["duration_minutes"]
    .agg(
        track_count="count",
        average_duration_min="mean",
        median_duration_min="median",
    )
    .reset_index()
)

explicit_artist_duration
```

```python
explicit_analysis_summary = pd.Series(
    {
        "total_tracks": len(tracks),
        "explicit_tracks": tracks["explicit"].sum(),
        "non_explicit_tracks": (~tracks["explicit"]).sum(),
        "explicit_percentage": tracks["explicit"].mean() * 100,
        "non_explicit_percentage": (
            (~tracks["explicit"]).mean() * 100
        ),
        "average_explicit_duration_min": tracks.loc[
            tracks["explicit"],
            "duration_minutes",
        ].mean(),
        "average_non_explicit_duration_min": tracks.loc[
            ~tracks["explicit"],
            "duration_minutes",
        ].mean(),
    }
)

explicit_analysis_summary
```

## 12. Data visualizations

1. Quantidade de tracks por artista

```python
plt.figure(figsize=(10, 6))

plt.hist(
    tracks["duration_minutes"],
    bins=30,
    edgecolor="black",
)

plt.axvline(
    tracks["duration_minutes"].mean(),
    linestyle="--",
    label="Mean",
)

plt.axvline(
    tracks["duration_minutes"].median(),
    linestyle=":",
    label="Median",
)

plt.title("Distribution of Track Duration")
plt.xlabel("Duration (minutes)")
plt.ylabel("Number of Tracks")

plt.legend()
plt.tight_layout()
plt.show()
```

## 13. Key insights

### Key findings

- Linkin Park has the largest number of tracks in the dataset, while Dire Straits has the highest average track duration.
- Pink Floyd has significantly fewer tracks than Linkin Park but a very similar total listening duration, reflecting its longer songs.
- Most tracks are concentrated between 3 and 5 minutes.
- The duration distribution is right-skewed due to a smaller number of very long tracks.
- Eminem has by far the highest proportion of explicit tracks among the analyzed artists.
- The global difference in duration between explicit and non-explicit tracks is strongly influenced by the artist composition of the dataset.
- Album size varies considerably, especially because expanded and anniversary editions contain substantially more tracks.
- Extremely short and long tracks identified as statistical outliers appear to represent legitimate musical content such as skits, intros, interludes, and extended compositions rather than data-quality errors.
