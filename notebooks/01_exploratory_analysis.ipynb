## 7. Descriptive statistics

```python
tracks[
    [
        "disc_number",
        "track_number",
        "duration_ms",
        "duration_seconds",
        "duration_minutes",
    ]
].describe()
```

```python
duration_summary = pd.Series(
    {
        "average_minutes": tracks["duration_minutes"].mean(),
        "median_minutes": tracks["duration_minutes"].median(),
        "minimum_minutes": tracks["duration_minutes"].min(),
        "maximum_minutes": tracks["duration_minutes"].max(),
    }
)

duration_summary
```

```python
dataset_counts = pd.Series(
    {
        "artists": len(artists),
        "albums": len(albums),
        "tracks": len(tracks),
    }
)

dataset_counts
```

```python
explicit_counts = tracks["explicit"].value_counts()

explicit_counts
```

```python
explicit_percentage = (
    tracks["explicit"]
    .value_counts(normalize=True)
    .mul(100)
)

explicit_percentage
```

```python
albums_per_artist = (
    albums.groupby("artist_name")
    .size()
    .sort_values(ascending=False)
)

albums_per_artist
```

```python
tracks_per_artist = (
    tracks.groupby("artist_name")
    .size()
    .sort_values(ascending=False)
)

tracks_per_artist
```

```python
total_duration_per_artist = (
    tracks.groupby("artist_name")["duration_minutes"]
    .sum()
    .sort_values(ascending=False)
)

total_duration_per_artist
```

```python
overview = pd.Series(
    {
        "total_artists": len(artists),
        "total_albums": len(albums),
        "total_tracks": len(tracks),
        "average_track_duration_min": tracks["duration_minutes"].mean(),
        "median_track_duration_min": tracks["duration_minutes"].median(),
        "explicit_tracks": tracks["explicit"].sum(),
        "explicit_percentage": tracks["explicit"].mean() * 100,
    }
)

overview
```

## 8. Artist-level analysis

```python
artist_analysis = (
    tracks.groupby("artist_name")
    .agg(
        total_tracks=("spotify_id", "count"),
        average_duration_min=("duration_minutes", "mean"),
        median_duration_min=("duration_minutes", "median"),
        total_duration_min=("duration_minutes", "sum"),
        explicit_tracks=("explicit", "sum"),
    )
    .reset_index()
)
```

```python
albums_by_artist = (
    albums.groupby("artist_name")
    .size()
    .reset_index(name="total_albums")
)
```

```python
artist_analysis = artist_analysis.merge(
    albums_by_artist,
    on="artist_name",
    how="left",
)
```

```python
artist_analysis["explicit_percentage"] = (
    artist_analysis["explicit_tracks"]
    / artist_analysis["total_tracks"]
    * 100
)
```

```python
artist_analysis = artist_analysis[
    [
        "artist_name",
        "total_albums",
        "total_tracks",
        "average_duration_min",
        "median_duration_min",
        "total_duration_min",
        "explicit_tracks",
        "explicit_percentage",
    ]
]
```

```python
artist_analysis = artist_analysis.sort_values(
    by="total_tracks",
    ascending=False,
)

artist_analysis
```

```python
artist_analysis.sort_values(
    by="average_duration_min",
    ascending=False,
)
```

```python
artist_analysis.sort_values(
    by="explicit_percentage",
    ascending=False,
)
```

```python
artist_analysis.sort_values(
    by="total_duration_min",
    ascending=False,
)
```
